```python
import pandas as pd
import numpy as np
from collections import Counter
import re

# Загрузка датасета
df = pd.read_csv('train.csv')

# Основная информация о датасете
print("Основная информация о датасете")
print(f"Размер датасета: {df.shape}")
print("\n Количество пропусков в каждом столбце:")
print(df.isnull().sum())


# Процент выживаемости по классам
print("\n\n Процент выживаемости по классам")
survival_by_class = df.groupby('Pclass')['Survived'].agg(['count', 'sum'])
survival_by_class['survival_rate'] = survival_by_class['sum'] / survival_by_class['count'] * 100
print(survival_by_class)

#  Самые популярные имена
def extract_first_name(name):

    # Убираем титулы и фамилии
    name = re.sub(r'["(),]', '', name)
    parts = name.split()
    # Ищем не титулы
    for part in parts:
        if '.' not in part and part not in ['Mr', 'Mrs', 'Miss', 'Master', 'Dr', 'Rev', 'Mlle', 'Major', 'Col', 'Capt', 'Don', 'Jonkheer', 'Sir', 'Lady', 'Mme', 'Ms']:
            return part
    return parts[-1]

df['FirstName'] = df['Name'].apply(extract_first_name)

print("\n\n Самые популярные имена ")
# Самые популярные мужские имена
male_names = df[df['Sex'] == 'male']['FirstName']
female_names = df[df['Sex'] == 'female']['FirstName']

print("Самые популярные мужские имена:")
print(male_names.value_counts().head())
print("\nСамые популярные женские имена:")
print(female_names.value_counts().head())

#Самые популярные имена по классам
print("\n\n САмые популярные имена по классам ")
for pclass in sorted(df['Pclass'].unique()):
    print(f"\n Класс {pclass}:")
    class_data = df[df['Pclass'] == pclass]

    male_names_class = class_data[class_data['Sex'] == 'male']['FirstName']
    female_names_class = class_data[class_data['Sex'] == 'female']['FirstName']

    print(f"  Популярные мужские имена: {male_names_class.value_counts().head(1).index[0] if not male_names_class.empty else 'Нет данных'}")
    print(f"  Популярные женские имена: {female_names_class.value_counts().head(1).index[0] if not female_names_class.empty else 'Нет данных'}")

#Пассажиры старше 44 лет
print("\n\n  Пассажиры страше 44 лет")
elderly_passengers = df[df['Age'] > 44]

# Мужчины младше 44 лет
print("\n\n Младще 44 лет + мужчины")
young_males = df[(df['Age'] < 44) & (df['Sex'] == 'male')]

# Количество n-местных кабин
print("\n\n Кол-во n- местных кабин")
# Убираем пустые значения в Cabin
cabin_data = df.dropna(subset=['Cabin'])
# Считаем, сколько людей в каждой кабине
cabin_counts = cabin_data['Cabin'].value_counts()
# Считаем, сколько кабин каждого типа
n_place_cabins = cabin_counts.value_counts().sort_index()
print("Количество кабин с n пассажирами:")
for n, count in n_place_cabins.items():
    print(f"  {n}-местных кабин: {count}")
```
