```python


class Chocolate:
  def __init__(self, price: float, additive: str = None):

        if price <= 0:
             raise ValueError('цена батончика больше нуля')

        self.price = price
        self.additive = additive



  def show_my_chocolate(self):
            if self.additive:
                print(f'Шоколадный батончик, добавка: {self.additive}')
            else:
                print(f"Обычный шоколадный батончик")
  def __str__(self):
     return(f'Шоколадный батончик, {self.price} рублей')

try:
         chocolate1 = Chocolate(89.5,"изюм")
         chocolate2 = Chocolate(57.7)
         chocolate3 = Chocolate(120.0, "кокос")

         chocolate1.show_my_chocolate()
         print(chocolate1)
         chocolate2.show_my_chocolate()
         print(chocolate2)
         chocolate3.show_my_chocolate()
         print(chocolate3)

except ValueError as e:
        print(e)
```
