```python
import polars as pl

# 1. Считываем датасет из файла train.csv с помощью polars
df = pl.read_csv('train.csv')

# 2. Выводим основную информацию о датасете
print("\n Основная информация о датасете ")

# Информация о типах данных
print("\n Типы данных столбцов (dtypes):")
print(df.dtypes)

# Число пропусков
print("\n Количество пропущенных значений в каждом столбце:")
print(df.null_count())

# Основные статистики (describe)
print("\n Основные статистики для числовых столбцов (describe):")
print(df.describe())

# Средние значения для числовых столбцов
print("\n Средние значения для числовых столбцов (mean):")
numeric_cols = df.select(pl.col(pl.NUMERIC_DTYPES)).columns

# 3. Количество пассажиров каждого класса (Pclass)
print("\n Количество пассажиров каждого класса (Pclass)")
pclass_counts = df.get_column("Pclass").value_counts().sort(by="Pclass")
print(pclass_counts)

# 4. Количество выживших мужчин и женщин
print("\n Количество выживших мужчин и женщин")
survival_by_sex = df.groupby("Sex").agg([
    pl.col("Survived").sum().alias("Survived_Count"),
    pl.count().alias("Total_Count")
])
print(survival_by_sex)

# 5. Часть таблицы с пассажирами, возраст которых больше 44 лет
print("\n Пассажиры старше 44 лет ")
passengers_over_44 = df.filter(pl.col("Age") > 44)
print(f"Найдено {passengers_over_44.shape[0]} пассажиров старше 44 лет.")
print(passengers_over_44)

# Дополнительно: покажем только некоторые столбцы для лучшей читаемости
print("\n Пассажиры старше 44 лет")
passengers_over_44_selected = df.filter(pl.col("Age") > 44).select([
    "PassengerId", "Name", "Age", "Sex", "Survived", "Pclass"
])
print(passengers_over_44_selected)
```
